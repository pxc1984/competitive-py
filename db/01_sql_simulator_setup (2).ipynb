
### Создаём пользователя и базу данных

Если вы использовали **вариант A или B** (локальный PostgreSQL), выполните ячейку ниже, чтобы задать пароль пользователю `postgres` и создать базу `bank_course`.

Если вы использовали **вариант C (Docker)** — база `bank_course` создастся автоматически чуть позже средствами Python (пароль `postgres` уже задан переменной окружения при запуске контейнера), эту ячейку можно пропустить.


```python
# Только для macOS с локальным PostgreSQL (Homebrew)
import platform, subprocess, os

if platform.system() == "Darwin":
    # Получаем имя текущего пользователя macOS
    user = os.environ.get('USER')
    
    print(f"Найден пользователь macOS: {user}")
    print("Создаем базу данных bank_course...")
    
    # 1) Создаем БД от имени текущего пользователя (он уже суперпользователь в Homebrew PostgreSQL)
    !createdb bank_course 2>/dev/null || echo "База уже существует"
    
    # 2) Создаем роль postgres, если её нет (опционально)
    !psql -c "CREATE ROLE postgres WITH SUPERUSER LOGIN PASSWORD 'postgres';" 2>/dev/null || echo "Роль postgres уже существует"
    
    # 3) Пароль для существующего пользователя (на всякий случай)
    !psql -c "ALTER USER {user} PASSWORD 'postgres';" 2>/dev/null || echo "Пароль для {user} установлен"
    
    print("Готово! База bank_course создана.")
elif platform.system() == "Linux":
    # Ваш код для Linux...
    !sudo -u postgres psql -c "ALTER USER postgres PASSWORD 'postgres';" 2>/dev/null || \
     su postgres -c "psql -c \"ALTER USER postgres PASSWORD 'postgres';\"" 2>/dev/null || \
     psql -U postgres -c "ALTER USER postgres PASSWORD 'postgres';"
    !sudo -u postgres createdb bank_course 2>/dev/null || \
     su postgres -c "createdb bank_course" 2>/dev/null || \
     createdb -U postgres bank_course
else:
    print("Похоже, вы на Windows — используйте вариант C (Docker) выше.")
```

Нам понадобятся: `psycopg2` (драйвер PostgreSQL), `sqlalchemy` (для pandas и `%sql`-магии), `faker` (генерация реалистичных ФИО и т.п.), `pandas`, `matplotlib` (ER-диаграмма), `ipython-sql` (удобные `%%sql`-ячейки).

```python
import os
import psycopg2

user = os.environ.get('USER')
BASE = dict(host="localhost", port=5432, user=user, password="postgres")

# 1. Создаём базу bank_course, если её нет
conn = psycopg2.connect(dbname="postgres", **BASE)
conn.autocommit = True          # CREATE DATABASE нельзя выполнять внутри транзакции
cur = conn.cursor()
cur.execute("SELECT 1 FROM pg_database WHERE datname = 'bank_course';")
if cur.fetchone() is None:
    cur.execute("CREATE DATABASE bank_course;")
    print("База bank_course создана")
cur.close()
conn.close()

# 2. Подключаемся к bank_course и оставляем соединение открытым
conn = psycopg2.connect(dbname="bank_course", **BASE)
conn.autocommit = True          # в вашем SQL уже есть свои BEGIN; ... COMMIT;
cur = conn.cursor()
print("✅ Подключение к bank_course успешно!")
```


## Шаг 3. Создание схемы (DDL)


```python

cur.execute('''BEGIN;
CREATE SCHEMA databases_and_sql;
SET search_path = databases_and_sql, public;
SET TIME ZONE 'Europe/Moscow';
CREATE TABLE shop_customers (
 customer_id SERIAL PRIMARY KEY, first_name VARCHAR(100), last_name VARCHAR(100),
 email VARCHAR(255), phone_number VARCHAR(20), created_at TIMESTAMP);
CREATE TABLE shop_addresses (address_id SERIAL PRIMARY KEY, customer_id INT NOT NULL REFERENCES shop_customers, address TEXT);
CREATE TABLE shop_categories (category_id SERIAL PRIMARY KEY, name VARCHAR, parent_id INT REFERENCES shop_categories);
CREATE TABLE shop_products (product_id SERIAL PRIMARY KEY, name VARCHAR, description TEXT, price NUMERIC(12,2) CHECK(price>0), category_id INT REFERENCES shop_categories, created_at TIMESTAMP);
CREATE TABLE shop_orders (order_id SERIAL PRIMARY KEY, customer_id INT NOT NULL REFERENCES shop_customers,
 order_date DATE, order_status VARCHAR CHECK(order_status IN ('Pending','Processing','Completed','Refunded','Cancelled')),
 delivery_date DATE, total_amount NUMERIC(12,2) CHECK(total_amount>0), shipping_address_id INT NOT NULL REFERENCES shop_addresses,
 CHECK(delivery_date IS NULL OR delivery_date>=order_date));
CREATE TABLE shop_orderitems (order_item_id SERIAL PRIMARY KEY, order_id INT NOT NULL REFERENCES shop_orders, product_id INT NOT NULL REFERENCES shop_products);
CREATE TABLE shop_payments (payment_id SERIAL PRIMARY KEY, order_id INT NOT NULL REFERENCES shop_orders, payment_date TIMESTAMP, amount NUMERIC(12,2), payment_method VARCHAR, payment_status VARCHAR CHECK(payment_status IN ('Success','Failed','Refunded')));
CREATE TABLE shop_reviews (review_id SERIAL PRIMARY KEY, product_id INT REFERENCES shop_products, customer_id INT REFERENCES shop_customers, comment TEXT, rating INT NOT NULL CHECK(rating BETWEEN 1 AND 5), created_at DATE);
CREATE TABLE flights_aircrafts_data (aircraft_code VARCHAR PRIMARY KEY, range INT CHECK(range>0), model_name VARCHAR);
CREATE TABLE flights_airports_data (airport_code VARCHAR PRIMARY KEY, coordinates POINT, timezone VARCHAR, city_name VARCHAR, airport_name_en VARCHAR);
CREATE TABLE flights_bookings (book_ref VARCHAR PRIMARY KEY, book_date TIMESTAMPTZ, total_amount NUMERIC(12,2));
CREATE TABLE flights_flights (flight_id INT PRIMARY KEY, flight_no VARCHAR, scheduled_departure TIMESTAMPTZ, scheduled_arrival TIMESTAMPTZ,
 departure_airport VARCHAR REFERENCES flights_airports_data, arrival_airport VARCHAR REFERENCES flights_airports_data,
 status VARCHAR CHECK(status IN ('Scheduled','On Time','Delayed','Departed','Arrived','Cancelled')),
 aircraft_code VARCHAR REFERENCES flights_aircrafts_data, actual_departure TIMESTAMPTZ, actual_arrival TIMESTAMPTZ,
 CHECK(scheduled_arrival>scheduled_departure), CHECK(actual_arrival IS NULL OR (actual_departure IS NOT NULL AND actual_arrival>actual_departure)));
CREATE TABLE flights_seats (aircraft_code VARCHAR REFERENCES flights_aircrafts_data, seat_no VARCHAR, fare_conditions VARCHAR CHECK(fare_conditions IN ('Economy','Comfort','Business')), PRIMARY KEY(aircraft_code,seat_no));
CREATE TABLE flights_tickets (ticket_no VARCHAR PRIMARY KEY, book_ref VARCHAR REFERENCES flights_bookings, passenger_id VARCHAR, passenger_name VARCHAR, contact_data JSONB);
CREATE TABLE flights_ticket_flights (ticket_no VARCHAR REFERENCES flights_tickets, flight_id INT REFERENCES flights_flights, fare_conditions VARCHAR CHECK(fare_conditions IN ('Economy','Comfort','Business')), amount NUMERIC(12,2) CHECK(amount>=0), PRIMARY KEY(ticket_no,flight_id));
CREATE TABLE flights_boarding_passes (ticket_no VARCHAR, flight_id INT, boarding_no INT, seat_no VARCHAR,
 PRIMARY KEY(ticket_no,flight_id), FOREIGN KEY(ticket_no,flight_id) REFERENCES flights_ticket_flights,
 UNIQUE(flight_id,boarding_no), UNIQUE(flight_id,seat_no));

CREATE VIEW fl_aircrafts_data AS SELECT * FROM flights_aircrafts_data;
CREATE VIEW fl_airports_data AS SELECT * FROM flights_airports_data;
CREATE VIEW fl_bookings AS SELECT * FROM flights_bookings;
CREATE VIEW fl_flights AS SELECT * FROM flights_flights;
CREATE VIEW fl_seats AS SELECT * FROM flights_seats;
CREATE VIEW fl_tickets AS SELECT * FROM flights_tickets;
CREATE VIEW fl_ticket_flights AS SELECT * FROM flights_ticket_flights;
CREATE VIEW fl_boarding_passes AS SELECT * FROM flights_boarding_passes;
CREATE INDEX ON shop_orders(customer_id,order_date);
CREATE INDEX ON shop_reviews(customer_id,created_at);
CREATE INDEX ON flights_flights(actual_departure);
CREATE INDEX ON flights_flights(scheduled_departure);
CREATE INDEX ON flights_ticket_flights(flight_id);
CREATE INDEX ON flights_tickets(book_ref);
CREATE VIEW orders AS SELECT order_id, customer_id AS customer, order_date, total_amount AS amount FROM shop_orders WHERE order_status='Completed';
COMMENT ON SCHEMA databases_and_sql IS 'Synthetic offline SQL seminar data; seed 6102026';
COMMIT;
''')
```


## Шаг 4. Вставка данных


```python
from sqlalchemy import create_engine
import pandas as pd

your_login = BASE["user"]        # = user (имя пользователя macOS)
your_pass = BASE["password"]     # = 'postgres'
host = BASE["host"]              # = 'localhost'
port = BASE["port"]              # = 5432
dbname = "bank_course"

engine = create_engine(
    f"postgresql+psycopg2://{your_login}:{your_pass}@{host}:{port}/{dbname}"
)

# Проверка, что подключился успешно
query = """
SELECT *
FROM databases_and_sql.shop_customers
LIMIT 5
"""
pd.read_sql(query, engine)
```
