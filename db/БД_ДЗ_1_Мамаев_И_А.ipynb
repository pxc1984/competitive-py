## Реляционная алгебра

### Задание 2

Получить список всех хостов `host_name` компании `TechCorp`.

Ответ:
π host_name (σ company_name='TechCorp' (host))


### Задание 3

Найти все успешные попытки подключения (отношение `connection_attempt`) к хосту с `host_id = 1`

Ответ:
σ host_id=1 ∧ status='success' (connection_attempt)


### Задание 4

Найти имена всех хостов и их IP-адреса. Результат должен содержать атрибуты `host_name` и `ip_address`.

Ответ:
π host_name, ip_address (host⨝ip_address)

### Задание 5

Получить список всех пользователей (атрибуты `login`, `name`), которые успешно подключались к хостам компании `DataSystems`. Исключи дубликаты.

Ответ:
π login, name (σ company_name='DataSystems' ∧ status='success' (connection_attempt⨝user_info⨝host))

### Задание 6

Получить список всех компаний (атрибут `company_name`) и количество хостов, зарегистрированных для каждой компании.

Ответ:
γ company_name; count(host_id)→host_count (host)

### Задание 7

Для каждого пользователя и компании найти количество успешных подключений к хостам этой компании. Результат должен содержать название компании, логин пользователя и количество подключений.

Ответ:
γ company_name, login; count(status)→connections (σ status='success' (host⨝user_info⨝connection_attempt))
