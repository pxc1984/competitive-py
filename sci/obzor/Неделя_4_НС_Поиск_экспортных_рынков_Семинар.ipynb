## Напоминание: КПВ и сравнительные преимущества

Две страны — **Альфа** и **Бета** — производят **мороженое (M)** и **кофе (K)**. За рабочий день один работник может произвести:

| Страна | Мороженое (M), тыс. порций | Кофе (K), тыс. чашек |
| ------ | --------------------------- | --------------------- |
| Альфа  | 10                           | 5                      |
| Бета   | 6                            | 6                      |

**Альтернативные издержки:**

| Страна | Издержки 1 ед. M (в K) | Издержки 1 ед. K (в M) |
| ------ | ------------------------ | ------------------------ |
| Альфа  | $0{,}5$                  | $2$                       |
| Бета   | $1$                       | $1$                       |

**Абсолютное преимущество:** Альфа — в мороженом (10 > 6), Бета — в кофе (6 > 5).
**Сравнительное преимущество:** издержки мороженого ниже в Альфе ($0{,}5<1$), издержки кофе ниже в Бете ($1<2$) $\Rightarrow$ Альфа специализируется на мороженом, Бета — на кофе, и это выгодно обеим странам при любой мировой цене $p=P_K/P_M$ между их автаркийными издержками кофе, т.е. $1<p<2$.

## Задача 1. RCA мороженого: формула и пример расчёта

Экспорт **мороженого (HS 2105)** четырьмя странами за иллюстративный год (млн долл.):

| Страна    | Экспорт мороженого, $X_{c,p}$ | Совокупный экспорт страны, $X_c$ |
| --------- | ------------------------------ | ---------------------------------- |
| Италия    | 500                             | 600 000                            |
| Германия  | 300                             | 1 500 000                          |
| Китай     | 150                             | 3 000 000                          |
| Россия    | 20                              | 450 000                            |
| Прочие    | 30                              | —                                   |

Мировой экспорт мороженого $X_p^w=1000$ млн долл., мировой совокупный экспорт $X^w=19\,000\,000$ млн долл.

**Задание.** Рассчитай RCA для Италии, Германии, Китая и России и определи, у кого есть выявленное сравнительное преимущество ($\text{RCA}\ge1$).

**Решение**

$$
\text{RCA}_{c,p} = \frac{X_{c,p}/X_c}{X_p^{w}/X^{w}}.
$$

Если $\text{RCA}>1$ — товар является выявленной специализацией страны, если $\text{RCA}<1$ — не является, а $\text{RCA}=1$ соответствует нейтральной позиции, при которой структура экспорта страны в точности повторяет мировую.

### 4.3 ECI и PCI через собственные векторы

Метод отражений сворачивается в страновую матрицу
$$
\tilde M = D_c^{-1} M D_p^{-1} M^{\top},
$$
где $D_c=\text{diag}(k_{c,0})$, $D_p=\text{diag}(k_{p,0})$. Первый собственный вектор $\tilde M$ тривиален (постоянный, не различает страны) — нас интересует **второй**: его координаты — это устойчивые значения $k_c$, из которых после нормировки получается
$$
\text{ECI}_c=\frac{k_c-\langle k_c\rangle}{\sigma(k_c)}.
$$
Симметрично для товаров строится $\tilde M^{(p)}=D_p^{-1}M^{\top}D_c^{-1}M$, второй собственный вектор которой даёт PCI.
